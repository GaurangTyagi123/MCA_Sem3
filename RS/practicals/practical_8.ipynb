# Practical 8: Graph-Based Recommender System Using GNN

## Aim

To model MovieLens Small as a **bipartite user-movie interaction graph**, implement a **LightGCN-style graph neural recommender**, generate Top-N recommendations, evaluate ranking performance, and analyze user-item interaction patterns.

The graph contains:

- user nodes
- movie nodes
- positive user-movie interaction edges

The propagation rule is:

\[
E^{(l+1)} = D^{-1/2} A D^{-1/2} E^{(l)}
\]

The model is trained with Bayesian Personalized Ranking (BPR) so that a user's observed positive movie receives a higher score than an unobserved negative movie.

## 17. Build normalized bipartite adjacency matrix


```python
# 17. Build normalized bipartite adjacency matrix
#
# Node indexing:
#
# user nodes:
# 0 ... NUM_USERS-1
#
# movie nodes:
# NUM_USERS ... NUM_USERS+NUM_MOVIES-1
#
# Each user-movie edge is made bidirectional.
#
# LightGCN propagation:
#
# E^(l+1) = D^(-1/2) A D^(-1/2) E^(l)
# ------------------------------------------------------------

def build_normalized_adjacency(
    user_indices,
    movie_indices,
    num_users,
    num_movies,
    device
):
    """
    Construct a symmetric normalized sparse adjacency matrix.

    Parameters
    ----------
    user_indices:
        Integer user indices for all training edges.

    movie_indices:
        Integer movie indices for all training edges.

    num_users:
        Number of user nodes.

    num_movies:
        Number of movie nodes.

    device:
        CPU or CUDA device.

    Returns
    -------
    normalized_adj:
        Sparse PyTorch matrix with shape
        (num_users + num_movies, num_users + num_movies).
    """

    movie_nodes = (
        movie_indices
        + num_users
    )

    row = np.concatenate(
        [
            user_indices,
            movie_nodes
        ]
    )

    col = np.concatenate(
        [
            movie_nodes,
            user_indices
        ]
    )

    values = np.ones(
        len(
            row
        ),
        dtype=np.float32
    )

    degree = np.bincount(
        row,
        minlength=(
            num_users
            + num_movies
        )
    ).astype(
        np.float32
    )

    degree_inv_sqrt = np.zeros_like(
        degree
    )

    nonzero = degree > 0

    degree_inv_sqrt[
        nonzero
    ] = (
        1.0
        / np.sqrt(
            degree[
                nonzero
            ]
        )
    )

    normalized_values = (
        values
        * degree_inv_sqrt[
            row
        ]
        * degree_inv_sqrt[
            col
        ]
    )

    indices = torch.tensor(
        np.vstack(
            [
                row,
                col
            ]
        ),
        dtype=torch.long
    )

    values_tensor = torch.tensor(
        normalized_values,
        dtype=torch.float32
    )

    normalized_adj = torch.sparse_coo_tensor(
        indices,
        values_tensor,
        size=(
            num_users
            + num_movies,
            num_users
            + num_movies
        )
    )

    normalized_adj = (
        normalized_adj
        .coalesce()
        .to(
            device
        )
    )

    return normalized_adj


normalized_adj = build_normalized_adjacency(
    train_user_indices,
    train_movie_indices,
    NUM_USERS,
    NUM_MOVIES,
    DEVICE
)

print(
    "\nNormalized adjacency shape:",
    normalized_adj.shape
)

print(
    "Number of directed sparse entries:",
    normalized_adj._nnz()
)


# ------------------------------------------------------------
```

## 18. LightGCN model


```python
# 18. LightGCN model
#
# LightGCN removes feature transforms and nonlinearities.
# It keeps the graph-neighborhood aggregation that matters for
# collaborative recommendation.
# ------------------------------------------------------------

class LightGCN(nn.Module):

    def __init__(
        self,
        num_users,
        num_movies,
        embedding_dim=64,
        num_layers=2
    ):
        """
        Parameters
        ----------
        num_users:
            Number of user nodes.

        num_movies:
            Number of movie nodes.

        embedding_dim:
            Size of every trainable node embedding.

        num_layers:
            Number of graph propagation steps.
        """

        super().__init__()

        self.num_users = (
            num_users
        )

        self.num_movies = (
            num_movies
        )

        self.embedding_dim = (
            embedding_dim
        )

        self.num_layers = (
            num_layers
        )

        self.user_embedding = nn.Embedding(
            num_users,
            embedding_dim
        )

        self.movie_embedding = nn.Embedding(
            num_movies,
            embedding_dim
        )

        nn.init.normal_(
            self.user_embedding.weight,
            std=0.1
        )

        nn.init.normal_(
            self.movie_embedding.weight,
            std=0.1
        )


    def propagate(
        self,
        normalized_adj
    ):
        """
        Perform LightGCN neighborhood aggregation.

        Initial embeddings are layer 0.

        Each new layer is obtained by multiplying the normalized
        adjacency matrix by the previous node embeddings.

        The final representation is the mean of embeddings from
        layer 0 through layer L.
        """

        all_embeddings = torch.cat(
            [
                self.user_embedding.weight,
                self.movie_embedding.weight
            ],
            dim=0
        )

        layer_embeddings = [
            all_embeddings
        ]

        current = (
            all_embeddings
        )

        for layer in range(
            self.num_layers
        ):

            current = torch.sparse.mm(
                normalized_adj,
                current
            )

            layer_embeddings.append(
                current
            )

        final_embeddings = torch.mean(
            torch.stack(
                layer_embeddings,
                dim=0
            ),
            dim=0
        )

        final_user_embeddings = (
            final_embeddings[
                :self.num_users
            ]
        )

        final_movie_embeddings = (
            final_embeddings[
                self.num_users:
            ]
        )

        return (
            final_user_embeddings,
            final_movie_embeddings
        )


    def score(
        self,
        user_indices,
        movie_indices,
        normalized_adj
    ):
        """
        Dot-product preference score for user-movie pairs.
        """

        (
            user_embeddings,
            movie_embeddings
        ) = self.propagate(
            normalized_adj
        )

        selected_users = (
            user_embeddings[
                user_indices
            ]
        )

        selected_movies = (
            movie_embeddings[
                movie_indices
            ]
        )

        scores = torch.sum(
            selected_users
            * selected_movies,
            dim=1
        )

        return scores


# ------------------------------------------------------------
```

## 19. Instantiate model and optimizer


```python
# 19. Instantiate model and optimizer
# ------------------------------------------------------------

model = LightGCN(
    num_users=NUM_USERS,
    num_movies=NUM_MOVIES,
    embedding_dim=EMBEDDING_DIM,
    num_layers=NUM_LAYERS
).to(
    DEVICE
)

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=LEARNING_RATE,
    weight_decay=WEIGHT_DECAY
)

print("\n" + "=" * 70)
print("LIGHTGCN MODEL")
print("=" * 70)

print(model)


# ------------------------------------------------------------
```

## 20. Negative sampling


```python
# 20. Negative sampling
#
# For each positive (user, movie) edge, sample one movie that
# the user did not positively interact with in the training graph.
# ------------------------------------------------------------

def sample_negative_movies(
    user_batch,
    positive_sets,
    num_movies
):
    """
    Sample one negative movie for each user in user_batch.

    user_batch:
        NumPy array of user indices.

    positive_sets:
        Dictionary user_index -> set of positive training movies.

    num_movies:
        Total number of movie nodes.
    """

    negatives = np.empty(
        len(
            user_batch
        ),
        dtype=np.int64
    )

    for position, user_index in enumerate(
        user_batch
    ):

        negative = np.random.randint(
            0,
            num_movies
        )

        while negative in positive_sets[
            int(
                user_index
            )
        ]:

            negative = np.random.randint(
                0,
                num_movies
            )

        negatives[
            position
        ] = negative

    return negatives


# ------------------------------------------------------------
```

## 21. BPR loss


```python
# 21. BPR loss
#
# We want:
#
# score(user, positive_movie)
# >
# score(user, negative_movie)
#
# Loss:
#
# -log sigmoid(pos_score - neg_score)
# ------------------------------------------------------------

def bpr_loss(
    user_embeddings,
    positive_embeddings,
    negative_embeddings
):
    """
    Bayesian Personalized Ranking loss for one batch.
    """

    positive_scores = torch.sum(
        user_embeddings
        * positive_embeddings,
        dim=1
    )

    negative_scores = torch.sum(
        user_embeddings
        * negative_embeddings,
        dim=1
    )

    loss = -torch.mean(
        F.logsigmoid(
            positive_scores
            - negative_scores
        )
    )

    return loss


# ------------------------------------------------------------
```

## 22. Train LightGCN


```python
# 22. Train LightGCN
# ------------------------------------------------------------

train_edge_count = len(
    train_user_indices
)

training_loss_history = []

for epoch in range(
    1,
    EPOCHS + 1
):

    model.train()

    permutation = np.random.permutation(
        train_edge_count
    )

    epoch_loss = 0.0
    batch_count = 0

    for start in range(
        0,
        train_edge_count,
        BATCH_SIZE
    ):

        batch_positions = permutation[
            start:
            start + BATCH_SIZE
        ]

        user_batch_np = train_user_indices[
            batch_positions
        ]

        positive_batch_np = train_movie_indices[
            batch_positions
        ]

        negative_batch_np = sample_negative_movies(
            user_batch_np,
            train_positive_sets,
            NUM_MOVIES
        )

        user_batch = torch.tensor(
            user_batch_np,
            dtype=torch.long,
            device=DEVICE
        )

        positive_batch = torch.tensor(
            positive_batch_np,
            dtype=torch.long,
            device=DEVICE
        )

        negative_batch = torch.tensor(
            negative_batch_np,
            dtype=torch.long,
            device=DEVICE
        )

        optimizer.zero_grad()

        (
            propagated_users,
            propagated_movies
        ) = model.propagate(
            normalized_adj
        )

        user_embeddings = (
            propagated_users[
                user_batch
            ]
        )

        positive_embeddings = (
            propagated_movies[
                positive_batch
            ]
        )

        negative_embeddings = (
            propagated_movies[
                negative_batch
            ]
        )

        loss = bpr_loss(
            user_embeddings,
            positive_embeddings,
            negative_embeddings
        )

        loss.backward()

        optimizer.step()

        epoch_loss += float(
            loss.item()
        )

        batch_count += 1

    mean_epoch_loss = (
        epoch_loss
        / max(
            batch_count,
            1
        )
    )

    training_loss_history.append(
        mean_epoch_loss
    )

    print(
        f"Epoch {epoch:02d} | "
        f"BPR Loss = "
        f"{mean_epoch_loss:.6f}"
    )


# ------------------------------------------------------------
```

## 23. Plot training loss


```python
# 23. Plot training loss
# ------------------------------------------------------------

plt.figure(
    figsize=(8, 5)
)

plt.plot(
    range(
        1,
        len(
            training_loss_history
        ) + 1
    ),
    training_loss_history
)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "BPR Loss"
)

plt.title(
    "LightGCN Training Loss"
)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
```

```python
# 24. Get final propagated embeddings
# ------------------------------------------------------------

model.eval()

with torch.no_grad():

    (
        final_user_embeddings,
        final_movie_embeddings
    ) = model.propagate(
        normalized_adj
    )


# ------------------------------------------------------------
```

## 25. Recommendation function


```python
# 25. Recommendation function
# ------------------------------------------------------------

def recommend_movies(
    user_id,
    top_n=TOP_N
):
    """
    Rank unseen movies for one user using the dot product between
    the learned user embedding and all movie embeddings.
    """

    if user_id not in user_to_index:
        return pd.DataFrame()

    user_index = user_to_index[
        user_id
    ]

    with torch.no_grad():

        user_vector = (
            final_user_embeddings[
                user_index
            ]
        )

        scores = torch.mv(
            final_movie_embeddings,
            user_vector
        )

        scores = scores.detach().cpu().numpy()

    seen_movies = train_positive_sets[
        user_index
    ]

    if len(
        seen_movies
    ) > 0:

        seen_array = np.array(
            list(
                seen_movies
            ),
            dtype=int
        )

        scores[
            seen_array
        ] = -np.inf

    valid_indices = np.where(
        np.isfinite(
            scores
        )
    )[0]

    if len(
        valid_indices
    ) == 0:
        return pd.DataFrame()

    n_keep = min(
        top_n,
        len(
            valid_indices
        )
    )

    valid_scores = scores[
        valid_indices
    ]

    top_local = np.argpartition(
        valid_scores,
        -n_keep
    )[-n_keep:]

    top_movie_indices = (
        valid_indices[
            top_local
        ]
    )

    top_movie_indices = (
        top_movie_indices[
            np.argsort(
                scores[
                    top_movie_indices
                ]
            )[::-1]
        ]
    )

    rows = []

    for rank, movie_index in enumerate(
        top_movie_indices,
        start=1
    ):

        movie_id = index_to_movie[
            int(
                movie_index
            )
        ]

        title = (
            movie_titles.loc[
                movie_id
            ]
            if movie_id
            in movie_titles.index
            else str(
                movie_id
            )
        )

        rows.append({
            "Rank": rank,
            "movieId": movie_id,
            "Movie": title,
            "GNN Score": float(
                scores[
                    movie_index
                ]
            )
        })

    return pd.DataFrame(
        rows
    )


# ------------------------------------------------------------
```

## 28. Ranking helper


```python
# 28. Ranking helper
# ------------------------------------------------------------

def rank_target_movie(
    user_id,
    target_movie_id
):
    """
    Return the 1-based rank of the held-out target movie.
    """

    if (
        user_id not in user_to_index
        or target_movie_id not in movie_to_index
    ):
        return None

    user_index = user_to_index[
        user_id
    ]

    target_movie_index = movie_to_index[
        target_movie_id
    ]

    with torch.no_grad():

        user_vector = (
            final_user_embeddings[
                user_index
            ]
        )

        scores = torch.mv(
            final_movie_embeddings,
            user_vector
        )

        scores = (
            scores
            .detach()
            .cpu()
            .numpy()
        )

    seen_movies = train_positive_sets[
        user_index
    ]

    if len(
        seen_movies
    ) > 0:

        seen_array = np.array(
            list(
                seen_movies
            ),
            dtype=int
        )

        scores[
            seen_array
        ] = -np.inf

    target_score = scores[
        target_movie_index
    ]

    if not np.isfinite(
        target_score
    ):
        return None

    rank = (
        int(
            np.sum(
                scores
                > target_score
            )
        )
        + 1
    )

    return rank


# ------------------------------------------------------------
```

## 29. Evaluate next positive item recommendation


```python
# 29. Evaluate next positive item recommendation
#
# Metrics:
# Hit Rate@10
# MRR@10
# NDCG@10
# ------------------------------------------------------------

evaluation_rows = []

for row in test_data.itertuples():

    rank = rank_target_movie(
        row.userId,
        row.movieId
    )

    hit = int(
        rank is not None
        and rank <= RANKING_K
    )

    rr = (
        1.0 / rank
        if hit
        else 0.0
    )

    ndcg = (
        1.0
        / np.log2(
            rank + 1
        )
        if hit
        else 0.0
    )

    evaluation_rows.append({
        "userId": row.userId,
        "movieId": row.movieId,
        "Rank": rank,
        f"Hit@{RANKING_K}": hit,
        f"RR@{RANKING_K}": rr,
        f"NDCG@{RANKING_K}": ndcg
    })


evaluation_results = pd.DataFrame(
    evaluation_rows
)

hit_rate = (
    evaluation_results[
        f"Hit@{RANKING_K}"
    ].mean()
)

mrr_at_k = (
    evaluation_results[
        f"RR@{RANKING_K}"
    ].mean()
)

ndcg_at_k = (
    evaluation_results[
        f"NDCG@{RANKING_K}"
    ].mean()
)

print("\n" + "=" * 70)
print("GNN RANKING EVALUATION")
print("=" * 70)

print(
    f"Hit Rate@{RANKING_K}: "
    f"{hit_rate:.4f}"
)

print(
    f"MRR@{RANKING_K}: "
    f"{mrr_at_k:.4f}"
)

print(
    f"NDCG@{RANKING_K}: "
    f"{ndcg_at_k:.4f}"
)

display(
    evaluation_results.head(15)
)


# ------------------------------------------------------------
```

```python
# 32. Visualize learned movie embeddings in 2D using SVD
#
# This is only for visualization.
# The recommender itself uses the full embedding dimension.
# ------------------------------------------------------------

movie_embedding_matrix = (
    final_movie_embeddings
    .detach()
    .cpu()
    .numpy()
)

centered_movie_embeddings = (
    movie_embedding_matrix
    - movie_embedding_matrix.mean(
        axis=0,
        keepdims=True
    )
)

U, S, Vt = np.linalg.svd(
    centered_movie_embeddings,
    full_matrices=False
)

movie_2d = (
    U[:, :2]
    * S[:2]
)

# Label only the most connected movies to avoid clutter.
label_movie_ids = (
    train_movie_degree
    .sort_values(
        ascending=False
    )
    .head(20)
    .index
    .tolist()
)

plt.figure(
    figsize=(9, 7)
)

plt.scatter(
    movie_2d[:, 0],
    movie_2d[:, 1],
    alpha=0.35
)

for movie_id in label_movie_ids:

    movie_index = movie_to_index[
        movie_id
    ]

    title = (
        movie_titles.loc[
            movie_id
        ]
        if movie_id
        in movie_titles.index
        else str(
            movie_id
        )
    )

    plt.annotate(
        title[:20],
        (
            movie_2d[
                movie_index,
                0
            ],
            movie_2d[
                movie_index,
                1
            ]
        ),
        fontsize=7
    )

plt.xlabel(
    "Embedding Dimension 1"
)

plt.ylabel(
    "Embedding Dimension 2"
)

plt.title(
    "2D View of Learned Movie Graph Embeddings"
)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
```
