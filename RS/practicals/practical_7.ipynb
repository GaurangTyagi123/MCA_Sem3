# Practical 7: Sequence-Aware Recommendation Using RNN

## Aim

To build a sequence-aware recommender system on MovieLens Small using a **SimpleRNN** and evaluate **next-item prediction** performance.

The model learns:

\[
P(i_{t+1} \mid i_1, i_2, \ldots, i_t)
\]

The practical uses:

- chronological user interaction sequences
- movie-token vocabulary
- sequence truncation and padding
- embedding layer
- SimpleRNN
- softmax next-item prediction
- Top-1 Accuracy
- Hit Rate@5
- Hit Rate@10
- MRR@10
- NDCG@10

## 5. Load MovieLens Small


```python
# 5. Load MovieLens Small
# ------------------------------------------------------------

ratings = pd.read_csv(
    RATINGS_FILE
)

movies = pd.read_csv(
    MOVIES_FILE
)

required_rating_columns = {
    "userId",
    "movieId",
    "rating",
    "timestamp"
}

required_movie_columns = {
    "movieId",
    "title",
    "genres"
}

if not required_rating_columns.issubset(
    ratings.columns
):
    raise ValueError(
        "ratings.csv must contain "
        "userId, movieId, rating and timestamp."
    )

if not required_movie_columns.issubset(
    movies.columns
):
    raise ValueError(
        "movies.csv must contain "
        "movieId, title and genres."
    )

print("=" * 70)
print("MOVIELENS SMALL")
print("=" * 70)

print(
    "Ratings shape:",
    ratings.shape
)

print(
    "Movies shape :",
    movies.shape
)

print(
    "Users:",
    ratings["userId"].nunique()
)

print(
    "Movies:",
    ratings["movieId"].nunique()
)

print(
    "Ratings:",
    len(ratings)
)

display(
    ratings.head()
)

display(
    movies.head()
)


# ------------------------------------------------------------
```

## 6. Sort interactions chronologically


```python
# 6. Sort interactions chronologically
# ------------------------------------------------------------

ratings_sorted = (
    ratings
    .sort_values(
        [
            "userId",
            "timestamp"
        ]
    )
    .reset_index(
        drop=True
    )
)

print(
    "\nChronologically sorted interactions:"
)

display(
    ratings_sorted.head(10)
)


# ------------------------------------------------------------
```

## 7. Build movie vocabulary


```python
# 7. Build movie vocabulary
#
# Token 0 is reserved for padding.
# Real movies begin from token 1.
# ------------------------------------------------------------

unique_movie_ids = np.sort(
    ratings_sorted[
        "movieId"
    ].unique()
)

movie_to_token = {
    movie_id: token
    for token, movie_id
    in enumerate(
        unique_movie_ids,
        start=1
    )
}

token_to_movie = {
    token: movie_id
    for movie_id, token
    in movie_to_token.items()
}

NUM_ITEMS = len(
    unique_movie_ids
)

VOCAB_SIZE = (
    NUM_ITEMS
    + 1
)

print(
    "\nNumber of modeled movies:",
    NUM_ITEMS
)

print(
    "Vocabulary size including padding:",
    VOCAB_SIZE
)


# ------------------------------------------------------------
```

## 8. Movie title lookup


```python
# 8. Movie title lookup
# ------------------------------------------------------------

movie_titles = (
    movies
    .drop_duplicates(
        "movieId"
    )
    .set_index(
        "movieId"
    )["title"]
)


# ------------------------------------------------------------
```

## 9. Build chronological user sequences


```python
# 9. Build chronological user sequences
# ------------------------------------------------------------

def build_user_sequences(
    ratings_frame
):
    """
    Convert timestamped interactions into ordered movie sequences.

    Parameters
    ----------
    ratings_frame:
        DataFrame containing userId, movieId and timestamp.

    Returns
    -------
    user_sequences:
        Dictionary:
        userId -> list of movie tokens in chronological order.
    """

    user_sequences = {}

    grouped = ratings_frame.groupby(
        "userId",
        sort=False
    )

    for user_id, group in grouped:

        movie_ids = (
            group[
                "movieId"
            ]
            .tolist()
        )

        tokens = [
            movie_to_token[
                movie_id
            ]
            for movie_id
            in movie_ids
        ]

        user_sequences[
            user_id
        ] = tokens

    return user_sequences


user_sequences = build_user_sequences(
    ratings_sorted
)

print(
    "\nNumber of user sequences:",
    len(
        user_sequences
    )
)


# ------------------------------------------------------------
```

## 10. Keep users with at least 3 interactions


```python
# 10. Keep users with at least 3 interactions
#
# We need:
# earlier interactions -> training
# second-last          -> validation target
# last                 -> test target
# ------------------------------------------------------------

user_sequences = {
    user_id: sequence
    for user_id, sequence
    in user_sequences.items()
    if len(sequence) >= 3
}

print(
    "Users with at least 3 interactions:",
    len(
        user_sequences
    )
)


# ------------------------------------------------------------
```

## 11. Show one user's temporal split


```python
# 11. Show one user's temporal split
# ------------------------------------------------------------

def token_sequence_to_titles(
    token_sequence
):
    """
    Convert a sequence of movie tokens to movie titles.
    """

    titles = []

    for token in token_sequence:

        movie_id = token_to_movie[
            token
        ]

        title = (
            movie_titles.loc[
                movie_id
            ]
            if movie_id
            in movie_titles.index
            else str(
                movie_id
            )
        )

        titles.append(
            title
        )

    return titles


if DEMO_USER_ID in user_sequences:

    demo_sequence = user_sequences[
        DEMO_USER_ID
    ]

else:

    DEMO_USER_ID = next(
        iter(
            user_sequences
        )
    )

    demo_sequence = user_sequences[
        DEMO_USER_ID
    ]


demo_train_sequence = (
    demo_sequence[:-2]
)

demo_validation_target = (
    demo_sequence[-2]
)

demo_test_target = (
    demo_sequence[-1]
)

print("\n" + "=" * 70)
print("EXAMPLE TEMPORAL SPLIT")
print("=" * 70)

print(
    "Demo user:",
    DEMO_USER_ID
)

print(
    "\nComplete sequence:"
)

for title in token_sequence_to_titles(
    demo_sequence
):
    print(
        " ->",
        title
    )

print(
    "\nTraining portion:"
)

for title in token_sequence_to_titles(
    demo_train_sequence
):
    print(
        " ->",
        title
    )

print(
    "\nValidation target:"
)

print(
    token_sequence_to_titles(
        [
            demo_validation_target
        ]
    )[0]
)

print(
    "\nTest target:"
)

print(
    token_sequence_to_titles(
        [
            demo_test_target
        ]
    )[0]
)


# ------------------------------------------------------------
```

## 12. Create training examples from prefixes


```python
# 12. Create training examples from prefixes
#
# Example:
#
# [A]       -> B
# [A, B]    -> C
# [A, B, C] -> D
#
# For long histories, only the most recent
# MAX_SEQUENCE_LENGTH tokens are retained.
# ------------------------------------------------------------

def create_training_examples(
    user_sequences,
    max_sequence_length=MAX_SEQUENCE_LENGTH
):
    """
    Create training input/target pairs.

    For each user:
        training portion = sequence[:-2]

    Every prefix predicts the next item.
    """

    input_sequences = []
    targets = []

    for user_id, sequence in (
        user_sequences.items()
    ):

        train_sequence = (
            sequence[:-2]
        )

        if len(
            train_sequence
        ) < 2:
            continue

        for end_position in range(
            1,
            len(
                train_sequence
            )
        ):

            prefix = (
                train_sequence[
                    :end_position
                ]
            )

            target = (
                train_sequence[
                    end_position
                ]
            )

            prefix = (
                prefix[
                    -max_sequence_length:
                ]
            )

            input_sequences.append(
                prefix
            )

            targets.append(
                target
            )

    return (
        input_sequences,
        np.array(
            targets,
            dtype=np.int32
        )
    )


train_sequences_raw, y_train = (
    create_training_examples(
        user_sequences
    )
)

print(
    "\nTraining examples:",
    len(
        train_sequences_raw
    )
)


# ------------------------------------------------------------
```

## 13. Pad training sequences


```python
# 13. Pad training sequences
#
# Pre-padding:
#
# [A, B, C]
#
# becomes:
#
# [0, 0, ..., A, B, C]
# ------------------------------------------------------------

X_train = pad_sequences(
    train_sequences_raw,
    maxlen=MAX_SEQUENCE_LENGTH,
    padding="pre",
    truncating="pre",
    value=0
)

print(
    "X_train shape:",
    X_train.shape
)

print(
    "y_train shape:",
    y_train.shape
)


# ------------------------------------------------------------
```

## 14. Build validation examples


```python
# 14. Build validation examples
#
# Input:
# all interactions before second-last item
#
# Target:
# second-last item
# ------------------------------------------------------------

def create_validation_data(
    user_sequences,
    max_sequence_length=MAX_SEQUENCE_LENGTH
):
    """
    Create one validation example per user.
    """

    sequences = []
    targets = []
    users = []

    for user_id, sequence in (
        user_sequences.items()
    ):

        input_sequence = (
            sequence[:-2]
        )

        target = (
            sequence[-2]
        )

        if len(
            input_sequence
        ) == 0:
            continue

        input_sequence = (
            input_sequence[
                -max_sequence_length:
            ]
        )

        sequences.append(
            input_sequence
        )

        targets.append(
            target
        )

        users.append(
            user_id
        )

    X_val = pad_sequences(
        sequences,
        maxlen=max_sequence_length,
        padding="pre",
        truncating="pre",
        value=0
    )

    y_val = np.array(
        targets,
        dtype=np.int32
    )

    return (
        X_val,
        y_val,
        users
    )


X_val, y_val, val_users = (
    create_validation_data(
        user_sequences
    )
)

print(
    "\nValidation examples:",
    len(
        y_val
    )
)


# ------------------------------------------------------------
```

## 15. Build the SimpleRNN model


```python
# 15. Build the SimpleRNN model
#
# Architecture:
#
# movie token
#    ↓
# Embedding
#    ↓
# SimpleRNN
#    ↓
# Dense + Softmax
#    ↓
# Probability of every possible next movie
# ------------------------------------------------------------

def build_rnn_model(
    vocab_size,
    embedding_dim=EMBEDDING_DIM,
    rnn_units=RNN_UNITS
):
    """
    Build a next-item prediction model.
    """

    model = Sequential([
        Embedding(
            input_dim=vocab_size,
            output_dim=embedding_dim,
            mask_zero=True
        ),

        SimpleRNN(
            rnn_units
        ),

        Dense(
            vocab_size,
            activation="softmax"
        )
    ])

    model.compile(
        optimizer="adam",
        loss="sparse_categorical_crossentropy",
        metrics=[
            "sparse_categorical_accuracy"
        ]
    )

    return model


model = build_rnn_model(
    vocab_size=VOCAB_SIZE
)

print("\n" + "=" * 70)
print("RNN MODEL")
print("=" * 70)

model.summary()


# ------------------------------------------------------------
```

## 16. Train the model


```python
# 16. Train the model
# ------------------------------------------------------------

history = model.fit(
    X_train,
    y_train,
    validation_data=(
        X_val,
        y_val
    ),
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    verbose=1
)


# ------------------------------------------------------------
```

## 18. Prepare one sequence for prediction


```python
# 18. Prepare one sequence for prediction
# ------------------------------------------------------------

def prepare_sequence(
    token_sequence,
    max_sequence_length=MAX_SEQUENCE_LENGTH
):
    """
    Truncate to the latest interactions and pad to fixed length.
    """

    recent = token_sequence[
        -max_sequence_length:
    ]

    prepared = pad_sequences(
        [
            recent
        ],
        maxlen=max_sequence_length,
        padding="pre",
        truncating="pre",
        value=0
    )

    return prepared


# ------------------------------------------------------------
```

## 19. Predict Top-N next items


```python
# 19. Predict Top-N next items
# ------------------------------------------------------------

def predict_next_items(
    model,
    history_tokens,
    top_n=TOP_N,
    exclude_seen=True
):
    """
    Predict Top-N next movies from one user's interaction history.

    Parameters
    ----------
    model:
        Trained Keras model.

    history_tokens:
        Chronological token sequence used as context.

    top_n:
        Number of recommendations.

    exclude_seen:
        If True, movies already in the user's history are masked.

    Returns
    -------
    DataFrame with rank, movieId, title and probability.
    """

    prepared = prepare_sequence(
        history_tokens
    )

    probabilities = (
        model.predict(
            prepared,
            verbose=0
        )[0]
    )

    # Token 0 is padding, never a recommendation.
    probabilities[0] = -np.inf

    if exclude_seen:

        seen_tokens = np.unique(
            history_tokens
        )

        probabilities[
            seen_tokens
        ] = -np.inf

    valid_tokens = np.where(
        np.isfinite(
            probabilities
        )
    )[0]

    n_keep = min(
        top_n,
        len(
            valid_tokens
        )
    )

    valid_scores = probabilities[
        valid_tokens
    ]

    top_local = np.argpartition(
        valid_scores,
        -n_keep
    )[-n_keep:]

    top_tokens = valid_tokens[
        top_local
    ]

    top_tokens = top_tokens[
        np.argsort(
            probabilities[
                top_tokens
            ]
        )[::-1]
    ]

    rows = []

    for rank, token in enumerate(
        top_tokens,
        start=1
    ):

        movie_id = token_to_movie[
            int(
                token
            )
        ]

        title = (
            movie_titles.loc[
                movie_id
            ]
            if movie_id
            in movie_titles.index
            else str(
                movie_id
            )
        )

        rows.append({
            "Rank": rank,
            "movieId": movie_id,
            "Movie": title,
            "Probability": float(
                probabilities[
                    token
                ]
            )
        })

    return pd.DataFrame(
        rows
    )


# ------------------------------------------------------------
```

## 22. Create test contexts


```python
# 22. Create test contexts
#
# Input:
# every interaction before the last item
#
# Target:
# last item
# ------------------------------------------------------------

def create_test_data(
    user_sequences,
    max_sequence_length=MAX_SEQUENCE_LENGTH
):
    """
    Create one next-item test example per user.
    """

    sequences = []
    targets = []
    users = []
    full_histories = []

    for user_id, sequence in (
        user_sequences.items()
    ):

        history_tokens = (
            sequence[:-1]
        )

        target = (
            sequence[-1]
        )

        if len(
            history_tokens
        ) == 0:
            continue

        input_sequence = (
            history_tokens[
                -max_sequence_length:
            ]
        )

        sequences.append(
            input_sequence
        )

        targets.append(
            target
        )

        users.append(
            user_id
        )

        full_histories.append(
            history_tokens
        )

    X_test = pad_sequences(
        sequences,
        maxlen=max_sequence_length,
        padding="pre",
        truncating="pre",
        value=0
    )

    y_test = np.array(
        targets,
        dtype=np.int32
    )

    return (
        X_test,
        y_test,
        users,
        full_histories
    )


(
    X_test,
    y_test,
    test_users,
    test_histories
) = create_test_data(
    user_sequences
)

print(
    "\nTest examples:",
    len(
        y_test
    )
)


# ------------------------------------------------------------
```

## 23. Predict probability distributions for all test users


```python
# 23. Predict probability distributions for all test users
# ------------------------------------------------------------

test_probabilities = model.predict(
    X_test,
    batch_size=BATCH_SIZE,
    verbose=1
)


# ------------------------------------------------------------
```

## 24. Evaluate next-item prediction


```python
# 24. Evaluate next-item prediction
#
# Metrics:
# Top-1 Accuracy
# HR@5
# HR@10
# MRR@10
# NDCG@10
# ------------------------------------------------------------

def evaluate_next_item(
    probability_matrix,
    targets,
    histories,
    top_k_1=TOP_K_1,
    top_k_2=TOP_K_2
):
    """
    Evaluate true next-item ranks.

    Already-seen movie tokens are masked before ranking because
    MovieLens users normally rate a movie only once.
    """

    rows = []

    for index in range(
        len(
            targets
        )
    ):

        probabilities = (
            probability_matrix[
                index
            ]
            .copy()
        )

        target_token = int(
            targets[
                index
            ]
        )

        history_tokens = np.unique(
            histories[
                index
            ]
        )

        # Padding token cannot be recommended.
        probabilities[0] = -np.inf

        # Remove previously seen movies.
        probabilities[
            history_tokens
        ] = -np.inf

        target_score = probabilities[
            target_token
        ]

        if not np.isfinite(
            target_score
        ):
            rank = None

        else:

            rank = (
                int(
                    np.sum(
                        probabilities
                        > target_score
                    )
                )
                + 1
            )

        top1 = int(
            rank == 1
        )

        hit_5 = int(
            rank is not None
            and rank <= top_k_1
        )

        hit_10 = int(
            rank is not None
            and rank <= top_k_2
        )

        rr_10 = (
            1.0 / rank
            if hit_10
            else 0.0
        )

        ndcg_10 = (
            1.0
            / np.log2(
                rank + 1
            )
            if hit_10
            else 0.0
        )

        rows.append({
            "Target Token": target_token,
            "Rank": rank,
            "Top-1 Correct": top1,
            f"Hit@{top_k_1}": hit_5,
            f"Hit@{top_k_2}": hit_10,
            f"RR@{top_k_2}": rr_10,
            f"NDCG@{top_k_2}": ndcg_10
        })

    results = pd.DataFrame(
        rows
    )

    metrics = {
        "Top-1 Accuracy": results[
            "Top-1 Correct"
        ].mean(),

        f"Hit Rate@{top_k_1}": results[
            f"Hit@{top_k_1}"
        ].mean(),

        f"Hit Rate@{top_k_2}": results[
            f"Hit@{top_k_2}"
        ].mean(),

        f"MRR@{top_k_2}": results[
            f"RR@{top_k_2}"
        ].mean(),

        f"NDCG@{top_k_2}": results[
            f"NDCG@{top_k_2}"
        ].mean()
    }

    return (
        results,
        metrics
    )


evaluation_results, metrics = (
    evaluate_next_item(
        test_probabilities,
        y_test,
        test_histories
    )
)


# ------------------------------------------------------------
```
